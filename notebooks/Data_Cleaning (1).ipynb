## **Data Cleaning**

### **1.Changing str dates to datetime**

```python
medical_df["Admission_Date"] = pd.to_datetime(
    medical_df["Admission_Date"],
    format="%m/%d/%Y"
)
```

```python
medical_df[["Admission_Date"]].head()  #YYYY-MM-DD this is the date format
```

```python
medical_df["Discharge_Date"] = pd.to_datetime(
    medical_df["Discharge_Date"],
    format="%m/%d/%Y"
)
```

```python
medical_df[["Discharge_Date"]].head()
```

```python
medical_df.info()
```

### **2.Converting contact_number to a string**

```python
medical_df["contact_number"] = medical_df["contact_number"].astype(str)
```

```python
medical_df["contact_number"].dtype
```

### **3.Duplicate records**

```python
medical_df.duplicated().sum()
```

### **4.Missing values**

```python
medical_df.isnull().sum()
```

```python
medical_df["Paid_Amount"] <= medical_df["Total_Amount"]
```
